```python
#Ejercicio 2
#Manipulación básica con pandas
import pandas as pd
data = {'NOMBRE':['Juan','Andrés', 'Mariana','Laura'],'EDADES':['23','18','32','28'],'CIUDAD':['Buenavenura','Restrepo','Cali','Valledupar']}
datos = pd.DataFrame(data)
print(datos)
#crear un filtro de las personas con edad igual o mayor a 25
filtro_edad = datos['EDADES'] >= '25'
personas_mayores_25 = datos[filtro_edad]
print(personas_mayores_25)
#crear una columa llamada categoría
datos['CATEGORIA'] = ['ADULTO' if edad >= '25' else 'MENOR' for edad in datos['EDADES']]
print(datos)
```

```python
#EJERCICIO 3
#Operación básica con NumPy
import numpy as np
array = np.array([1,2,3,4,5])
print(array)
#Calcula la media y la desviación estándar del array vector.
media = np.mean(array)
desviacion_estandar = np.std(array)
print(media)
print(desviacion_estandar)
# Crea una matriz llamada matriz con 3 filas y 2 columnas, llenándola con números aleatorios entre 1 y 10.
matriz = np.random.randint(1,11,size=(3,2))
print(matriz)
#Multiplica todos los elementos de la matriz por 2.
matriz_multiplicada = matriz * 2
print(matriz_multiplicada)
#Calcula la suma de todos los elementos de la matriz.
suma_matriz = np.sum(matriz)
print(suma_matriz)
```

```python
# EJERCICIO 5
#Concatenación de DataFrames: Crea un programa que utilice Pandas para realizar las siguientes tareas:

#Crea dos DataFrames llamados df1 y df2 con las siguientes columnas: A, B, C.
#Llena ambos DataFrames con datos.
#Concatena verticalmente los dos DataFrames para crear un nuevo DataFrame llamado df_concat.
#Muestra por pantalla el resultado de la concatenación.
import pandas as pd
df1 = pd.DataFrame({'A':[1,2,3],'B':[4,5,6],'C':[7,8,9]})
df2 = pd.DataFrame({'A':[10,11,12],'B':[13,14,15],'C':[16,17,18]})
print(df1)
print(df2)
df_concat = pd.concat([df1, df2], ignore_index=True)
print(df_concat)
```
