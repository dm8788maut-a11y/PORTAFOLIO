```python
#EJERCICIO 4
#Tornillo: Cree un programa que solicite el tamaño de un tornillo e imprima su tamaño de acuerdo a las siguientes condiciones: De 1 cm(incluído) hasta 3 cm(no incluído) es pequeño
#De 3 cm(incluído) hasta 5 cm(no incluído) es mediano
#De 5 cm(incluído) hasta 6.5 cm(no incluído) es grande
#De 6.5cm (incluído) hasta 8.5 cm(no incluído) es muy grande
#De 8.5 cm(incluído) en adelante es gigante

try:
    tamano_tornillo = float(input("Ingrese el tamaño del tornillo en centímetros: "))
    if tamano_tornillo >= 1 and tamano_tornillo < 3:
        print("El tornillo es pequeño.")
    elif tamano_tornillo >= 3 and tamano_tornillo < 5:
        print("El tornillo es mediano.")
    elif tamano_tornillo >= 5 and tamano_tornillo < 6.5:
        print("El tornillo es grande.")
    elif tamano_tornillo >= 6.5 and tamano_tornillo < 8.5:
        print("El tornillo es muy grande.")
    elif tamano_tornillo >= 8.5:
        print("El tornillo es gigante.")
    else:
        print("El tamaño ingresado no corresponde a ninguna categoría.")
except ValueError:
    print("El valor ingresado no es un número válido.")
```
