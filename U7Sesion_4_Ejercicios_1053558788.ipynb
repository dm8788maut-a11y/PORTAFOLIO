```python
#Ejercicio 1
#Materias: Escribir un programa que almacene las asignaturas de un curso (por ejemplo Matemáticas, Física, Química, Historia, Lenguaje y Programación) en una lista, pregunte al usuario la nota que ha sacado en cada asignatura, y después las muestre por pantalla con el mensaje En <asignatura> has sacado <nota> dónde <asignatura> es cada una des las asignaturas de la lista y <nota> cada una de las correspondientes notas introducidas por el usuario.
asignaturas = ["Matemáticas", "Física", "Química", "Historia", "Lenguaje", "Programación"]
notas = []
for asignatura in asignaturas:
  try:
    nota_str = input(f"¿Qué nota has sacado en {asignatura}?: ")
    nota = float(nota_str)
    notas.append(nota)
  except ValueError:
    print("Por favor, introduce una nota válida.")

for i in range(len(asignaturas)):
  print(f"En {asignaturas[i]} has sacado {notas[i]}")
```

```python
#Ejercicio 2
#Cesta de compras: Escribir un programa que cree un diccionario simulando una cesta de la compra. El programa debe preguntar el artículo y su precio y añadir el par al diccionario, hasta que el usuario decida terminar ingresando la palabra “FIN”. Después se debe mostrar por pantalla la lista de la compra y el coste total, de la siguiente manera

#Articulo 1         Valor
#Articulo 2         Valor
#Articulo 3         Valor
#TOTAL        VALOR

Cesta = {}
total_compra = 0.0
print("cesta de compras")
while True:
  articulo = input("Introduce el nombre del artículo (o 'FIN' para terminar): ")
  if articulo.upper() == "FIN": # Use .upper() to handle both "FIN" and "fin"
    break
  while True: # Corrected 'true' to 'True'
    precio_input = input(f"Introduce el precio de {articulo}': ") # Corrected variable name to precio_input
    try:
      precio = float(precio_input)
      Cesta[articulo] = precio # Add the article and price to the dictionary
      total_compra += precio
      break # Break out of the inner while loop after getting a valid price
    except ValueError:
      print("Por favor, introduce un precio válido.")

# Display the shopping list and total cost
print("\n--- Lista de la Compra ---")
for articulo, precio in Cesta.items():
  print(f"{articulo:<15} {precio:>.2f}") # Format output for better alignment

print(f"\nTOTAL           {total_compra:>.2f}") # Format total cost
```

```python
#Ejercicio 3
#Gestiona una base de datos de clientes con un menú interactivo.
clientes = {}
while True:
  print("\n---MENU DE GESTION DE CLIENTES---")
  print("1 añadir cliente")
  print("2 eliminar cliente")
  print("3 mostrar cliente")
  print("4 listar todos los clientes")
  print("5 mostar clientes preferenciales")
  print("6 salir")
  opcion = input("Elige una opcion (1-6): ")
  if opcion == '1':
    nit = input("Introduce el nit del cliente: ")
    nombre = input("Introduce el nombre del cliente: ")
    email = input("Introduce el email del cliente: ")
    direccion = input("Introduce la direccion del cliente: ")
    preferencial_input=input("¿es preferencial (s/n): ").lower()
    preferencial = preferencial_input == 's'
    clientes[nit] = {'nombre':nombre,
                     'direccion':direccion,
                     'email':email,
                     'preferencial':preferencial}
    print("Cliente con NIT añadido correctamente")
  elif opcion == '5':
    break
  else:
    print("Opcion no valida. Por favor, elige una opcion del 1 al 5.")
  if opcion == '2':
    nit = input("Introduce el nit del cliente a eliminar: ")
    if nit in clientes:
      del clientes[nit]
      print("Cliente con NIT eliminado correctamente")
    else:
      print("No existe un cliente con ese NIT")
  if opcion == '3':
    nit = input("Introduce el nit del cliente a mostrar: ")
    if nit in clientes:
      print("Nombre:", clientes[nit]['nombre'])
      print("Direccion:", clientes[nit]['direccion'])
  if opcion == '4':
    for nit, datos in clientes.items():
      print("NIT:", nit)
      print("Nombre:", datos['nombre'])
      print("Direccion:", datos['direccion'])
  if opcion == '5':
    print("Clientes preferenciales:")
    for nit, datos in clientes.items():
      if datos['preferencial']:
        print("NIT:", nit)
        print("Nombre:", datos['nombre'])
        print("Direccion:", datos['direccion'])
  if opcion == '6':
    print("Saliendo del programa...")
    break
```

```python
#Ejercicio 4
# Programa para registrar empleados

empleados = []
while True:
  print("\n---MENU DE GESTION DE EMPLEADOS---")
  print("(1) añadir empleado")
  print("(2) mostrar todos los empleados")
  print("(3) salir")

  opcion = input("Elige una opcion (1-3): ")

  if opcion == '1':
    nombre = input("Introduce el nombre del empleado: ")
    salario = float(input("Introduce el salario del empleado: "))
    cargo = input("Introduce el cargo del empleado: ")
    nuevo_empleado = {'nombre': nombre, 'salario': salario, 'cargo': cargo}
    empleados.append(nuevo_empleado)
    print(f"Empleado {nombre} añadido correctamente")

  if opcion == '2':
    print("\n---LISTA DE EMPLEADOS REGISTRADOS---")
    for empleado in empleados:
      print(f"Nombre: {empleado['nombre']}")
      print(f"Salario: {empleado['salario']}")
  if opcion == '3':
      print("Saliendo del programa...")
      break
```

```python
#Ejercicio 5
# Promedio: Escribe un programa que solicite al usuario ingresar notas de estudiantes. El programa debe permitir al usuario ingresar cuantas notas desee. Cuando el usuario ingrese -1, el programa debe calcular y mostrar el promedio de las notas ingresadas.
suma_notas = 0
contador_notas = 0
while True:
  nota_str = input("Ingrese una nota (o -1 para calcular el promedio): ")
  if nota_str == "-1":
    if contador_notas == 0:
      print("No se ingresaron notas.")
    else:
      promedio = suma_notas / contador_notas
      print(f"El promedio de las notas ingresadas es: {promedio}")
    break
  try:
    nota = float(nota_str)
    suma_notas += nota
    contador_notas += 1
  except ValueError:
    print("Por favor, ingrese un número válido para la nota.")
```

```python
#ejercicio 6
#Adivinanza: Crea un programa que genere un número aleatorio entre 1 y 10 y permita al usuario intentar adivinarlo. El programa debe dar pistas indicando si el número ingresado es mayor o menor al número secreto. El juego debe continuar hasta que el usuario adivine el número.
import random
número_secreto = random.randint(1, 10)
adivinado = False
print("Juego de adivinanza")
print("Intenta adivinar el número entre 1 y 10")
while not adivinado:
  try:
    intento = int(input("Ingresa tu intento: "))
    if intento == número_secreto:
        print(f"¡Has adivinado el número, es el {número_secreto}!")
        adivinado = True
    elif intento < número_secreto:
        print("El número secreto es mayor. ¡Sigue intentando!")
    else:
        print("El número secreto es menor. ¡Sigue intentando!")
  except ValueError:
    print("Por favor, ingresa un número válido.")
```
